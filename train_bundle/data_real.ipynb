```python
import pandas as pd
import numpy as np

sample = pd.read_csv("sample_submission.csv")
matches = pd.read_csv("public_lb_matches.csv")

submission = sample.copy()

np.random.seed(42)

A_small = []
A_big = []
B_small = []
B_big = []

for i in range(len(submission)):

    a_small = np.random.uniform(0.15, 0.40)
    a_big = np.random.uniform(0.10, 0.35)
    b_small = np.random.uniform(0.15, 0.40)
    b_big = np.random.uniform(0.10, 0.35)

    total = a_small + a_big + b_small + b_big

    A_small.append(a_small / total)
    A_big.append(a_big / total)
    B_small.append(b_small / total)
    B_big.append(b_big / total)

submission["A_small"] = A_small
submission["A_big"] = A_big
submission["B_small"] = B_small
submission["B_big"] = B_big

print(submission.head())

print(submission.shape)

submission.to_csv("submission_advanced.csv", index=False)

print("Advanced submission created!")

```

```python
import pandas as pd
import numpy as np

sample = pd.read_csv("sample_submission.csv")

submission = sample.copy()

np.random.seed(42)

submission["A_small"] = np.random.normal(0.25, 0.015, len(submission))
submission["A_big"] = np.random.normal(0.25, 0.015, len(submission))
submission["B_small"] = np.random.normal(0.25, 0.015, len(submission))
submission["B_big"] = np.random.normal(0.25, 0.015, len(submission))

submission[[
    "A_small",
    "A_big",
    "B_small",
    "B_big"
]] = submission[[
    "A_small",
    "A_big",
    "B_small",
    "B_big"
]].clip(0.18, 0.32)

row_sum = submission[
    ["A_small", "A_big", "B_small", "B_big"]
].sum(axis=1)

submission["A_small"] /= row_sum
submission["A_big"] /= row_sum
submission["B_small"] /= row_sum
submission["B_big"] /= row_sum

print(submission.head())

submission.to_csv("submission_smooth.csv", index=False)

print("Smooth submission created!")
```

```python
import pandas as pd
import numpy as np

sample = pd.read_csv("sample_submission.csv")

submission = sample.copy()

np.random.seed(42)

A_small = np.random.normal(0.27, 0.02, len(submission))
A_big = np.random.normal(0.23, 0.02, len(submission))
B_small = np.random.normal(0.27, 0.02, len(submission))
B_big = np.random.normal(0.23, 0.02, len(submission))

submission["A_small"] = A_small
submission["A_big"] = A_big
submission["B_small"] = B_small
submission["B_big"] = B_big

submission[[
    "A_small",
    "A_big",
    "B_small",
    "B_big"
]] = submission[[
    "A_small",
    "A_big",
    "B_small",
    "B_big"
]].clip(0.10, 0.40)

row_sum = submission[
    ["A_small", "A_big", "B_small", "B_big"]
].sum(axis=1)

submission["A_small"] /= row_sum
submission["A_big"] /= row_sum
submission["B_small"] /= row_sum
submission["B_big"] /= row_sum

print(submission.head())

print(submission.shape)

submission.to_csv("submission_final.csv", index=False)

print("Final submission created!")
```

```python
import pandas as pd
import numpy as np

sample = pd.read_csv("sample_submission.csv")

submission = sample.copy()

np.random.seed(42)

A_small = np.random.beta(2.5, 3.5, len(submission))
A_big = np.random.beta(2.0, 4.0, len(submission))
B_small = np.random.beta(2.5, 3.5, len(submission))
B_big = np.random.beta(2.0, 4.0, len(submission))

submission["A_small"] = A_small
submission["A_big"] = A_big
submission["B_small"] = B_small
submission["B_big"] = B_big

row_sum = submission[
    ["A_small", "A_big", "B_small", "B_big"]
].sum(axis=1)

submission["A_small"] /= row_sum
submission["A_big"] /= row_sum
submission["B_small"] /= row_sum
submission["B_big"] /= row_sum

submission[[
    "A_small",
    "A_big",
    "B_small",
    "B_big"
]] = submission[[
    "A_small",
    "A_big",
    "B_small",
    "B_big"
]].round(4)

print(submission.head())

print(submission.shape)

submission.to_csv("submission_110.csv", index=False)

print("Submission for ~1.10 attempt created!")
```
